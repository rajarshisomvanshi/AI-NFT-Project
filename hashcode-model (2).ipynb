```python
from diffusers import StableDiffusionPipeline
import torch
from PIL import Image
import imagehash
import os

# --- Model Setup ---
# It's good practice to check if the model directory exists.
model_path = "/kaggle/working/stable-diffusion-v1-5"
if not os.path.isdir(model_path):
    print(f"Model path not found: {model_path}")
    print("Please ensure you have downloaded and saved the model correctly.")
    # As we can't proceed, we'll exit gracefully.
    # In a real application, you might download the model here.
    exit()

print("Loading the Stable Diffusion pipeline...")
pipe = StableDiffusionPipeline.from_pretrained(
    model_path,
    torch_dtype=torch.float16
).to("cuda")

pipe.safety_checker = None
pipe.enable_attention_slicing()
print("Pipeline loaded successfully.")


# --- Hash Persistence Functions ---

def load_hashes(hash_file_path):
    """Loads a set of hashes from a text file."""
    if not os.path.exists(hash_file_path):
        return set() # Return an empty set if the file doesn't exist
    
    print(f"Loading existing hashes from {hash_file_path}...")
    try:
        with open(hash_file_path, 'r') as f:
            # Create a set of imagehash objects from the strings in the file
            hashes = {imagehash.hex_to_hash(line.strip()) for line in f if line.strip()}
        print(f"Loaded {len(hashes)} unique hashes from history.")
        return hashes
    except Exception as e:
        print(f"Could not read hash file: {e}. Starting with an empty set.")
        return set()

def save_hash(hash_to_save, hash_file_path):
    """Appends a new hash to the hash file."""
    with open(hash_file_path, 'a') as f:
        f.write(str(hash_to_save) + "\n")



def generate_unique_images(
    user_prompt,
    num_unique_images=2,
    similarity_threshold=5,
    max_retries=10,
    output_prefix="/kaggle/working/nft_image",
    hash_log_file="image_hashes.log"
):
    
    style_suffix = ", 3D render, octane render, cinematic, hyper-detailed, intricate, volumetric lighting, unreal engine 5, trending on artstation, 8k"
    final_prompt = user_prompt + style_suffix
    print(f"\nUsing final prompt: {final_prompt}")

    seen_hashes = load_hashes(hash_log_file)
    initial_hash_count = len(seen_hashes)
    
    output_paths = []
    attempts = 0

    while len(output_paths) < num_unique_images:
        if attempts >= max_retries * num_unique_images:
            print(f"\nWarning: Reached max total attempts ({attempts}). Stopping early.")
            break

        print(f"\nGenerating image {len(output_paths) + 1} of {num_unique_images}...")
        attempts += 1

        image = pipe(final_prompt, num_inference_steps=25).images[0]

        current_hash = imagehash.phash(image)
        print(f"Generated a new image with hash: {current_hash}")

        is_duplicate = False
        if not seen_hashes:
            print("This is the first image ever, automatically considered unique.")
        else:
            for h in seen_hashes:
                distance = current_hash - h
                if distance <= similarity_threshold:
                    print(f"DUPLICATE DETECTED! Image is too similar to a previous one (distance: {distance}). Discarding.")
                    is_duplicate = True
                    break # No need to check other hashes

        if not is_duplicate:
            print(f"Image is unique. Saving image and its hash...")
            
            seen_hashes.add(current_hash)
            
            save_hash(current_hash, hash_log_file)

            path = f"{output_prefix}_{initial_hash_count + len(output_paths) + 1}.png"
            image.save(path)
            output_paths.append(path)

    return output_paths


if __name__ == "__main__":
    user_input = input("Enter your prompt for the NFT artwork: ")

    output_files = generate_unique_images(
        user_input, 
        num_unique_images=2, 
        similarity_threshold=5,
        hash_log_file="/kaggle/working/persistent_hashes.log"
    )

    print("\n Complete")
    if output_files:
        print("The following unique images were saved:")
        for path in output_files:
            print(f"- {path}")
    else:
        print("No unique images could be generated with the given prompt and settings.")
```
